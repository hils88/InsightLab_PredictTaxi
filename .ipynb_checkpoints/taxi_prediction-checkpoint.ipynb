```python
#importar clase creada en python
from analisis_taxi import AnalisisTaxi
```

```python
#instanciar
at = AnalisisTaxi()
```

```python
#carga de datos de taxi
at.cargar_datos("yellow_tripdata_2016-03.csv", nrows=200000)
```

```python
#realizar limpieza de datos
at.limpiar()
```

```python
#filtrar datos que esten fuera de rango
at.filtrar_outliers()
```

```python
# despues de una ejecucion, se filtran mas outliers
# Duración entre 1 y 60 min, por ejemplo
at.df = at.df[(at.df['duration_minutes'] >= 1) & (at.df['duration_minutes'] <= 60)]
print("Nuevo tamaño de dataset:", len(at.df))
```

```python
#preparacion de datos
at.preparar_train_test(target='duration_minutes', sample_frac=0.3)
```

```python
#crear y entrenar modelo de distancia
at.crear_pipeline_random_forest(n_estimators=150, max_depth=15,min_samples_leaf=5)
```

```python
# Crear y entrenar modelo de costo
at.crear_pipeline_random_forest_costo(n_estimators=150, max_depth=15, min_samples_leaf=5)
```

```python
#entrenar
at.entrenar()
```

```python
#evaluacion
metrics = at.evaluar()
```

```python
at.validacion_cruzada(cv=3)
```

```python
#prediccion de ejemplo viaje nuevo
row = {
    'trip_distance': 3.2,
    'pickup_hour': 15,
    'pickup_dow': 4,
    'is_weekend': 0,
    'is_peak': 1,
    'pickup_location_id': '237',
    'dropoff_location_id': '161'
}
pred_duracion = at.predecir_fila(row)
print("Pred duration (min):", round(pred_duracion, 2))
```

```python
#Graficos para ver la prediccion
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# Histograma de duracion
plt.figure(figsize=(8,5))
plt.hist(at.df['duration_minutes'], bins=50, color='skyblue', alpha=0.7)
plt.axvline(pred_duracion, color='red', linestyle='dashed', linewidth=2, label='Prediccion')
plt.xlabel('Duracion (min)')
plt.ylabel('Cantidad de viajes')
plt.title('Distribución de Duracion de Viajes')
plt.legend()
plt.show()

# Prediccion según distancia
distancias = np.linspace(0.5, 10, 20)
preds = [at.predecir_fila({'trip_distance': d,
                           'pickup_hour': 15,
                           'pickup_dow': 4,
                           'is_weekend': 0,
                           'is_peak': 1,
                           'pickup_location_id': '237',
                           'dropoff_location_id': '161'}) 
         for d in distancias]

plt.figure(figsize=(8,5))
plt.plot(distancias, preds, marker='o', color='orange')
plt.xlabel('Distancia del viaje (mi)')
plt.ylabel('Duracion predicha (min)')
plt.title('Predicción de duración según distancia')
plt.grid(True)
plt.show()
```

```python
# Filtrar vecinos por distancia +/-0.5, hora +/-1 y mismas locs
df = at.df.copy()
mask = (
    (df['pickup_location_id'] == row['pickup_location_id']) &
    (df['dropoff_location_id'] == row['dropoff_location_id']) &
    (df['pickup_hour'].between(row['pickup_hour']-1, row['pickup_hour']+1)) &
    (df['trip_distance'].between(row['trip_distance']-0.5, row['trip_distance']+0.5))
)
neigh = df[mask]
print("Vecinos encontrados:", len(neigh))
if len(neigh) > 0:
    print("Duración mediana (min):", neigh['duration_minutes'].median())
    print("Duración promedio (min):", neigh['duration_minutes'].mean())
else:
    mask2 = (
        (df['trip_distance'].between(row['trip_distance']-1.0, row['trip_distance']+1.0)) &
        (df['pickup_hour'].between(row['pickup_hour']-2, row['pickup_hour']+2))
    )
    neigh2 = df[mask2]
    print("Vecinos sueltos:", len(neigh2))
    if len(neigh2)>0:
        print("Mediana (vecinos sueltos):", neigh2['duration_minutes'].median())
```

```python
mask2 = (
    (at.df['trip_distance'].between(row2['trip_distance']-1.0, row2['trip_distance']+1.0)) &
    (at.df['pickup_hour'].between(row2['pickup_hour']-2, row2['pickup_hour']+2))
)
neigh2 = at.df[mask2]
if len(neigh2) > 0:
    print(f"Vecinos encontrados viaje 2: {len(neigh2)}")
    print(f"Mediana duración vecinos viaje 2: {neigh2['duration_minutes'].median():.2f} min")
else:
    print("No se encontraron vecinos similares viaje 2.")
```

```python
#presentacion de viajes
viajes = [
    {'trip_distance': 3.2, 'pickup_hour': 15, 'pickup_dow': 4, 'is_weekend': 0, 'is_peak': 1, 'pickup_location_id': '237', 'dropoff_location_id': '161'},
    {'trip_distance': 7.5, 'pickup_hour': 9,  'pickup_dow': 2, 'is_weekend': 0, 'is_peak': 1, 'pickup_location_id': '100', 'dropoff_location_id': '230'},
    {'trip_distance': 1.8, 'pickup_hour': 22, 'pickup_dow': 5, 'is_weekend': 1, 'is_peak': 0, 'pickup_location_id': '50',  'dropoff_location_id': '12'},
    {'trip_distance': 12.0,'pickup_hour': 8,  'pickup_dow': 1, 'is_weekend': 0, 'is_peak': 1, 'pickup_location_id': '45',  'dropoff_location_id': '90'},
    {'trip_distance': 5.5, 'pickup_hour': 18, 'pickup_dow': 6, 'is_weekend': 1, 'is_peak': 1, 'pickup_location_id': '210', 'dropoff_location_id': '55'}
]

import pandas as pd
resultados = []

for i, row in enumerate(viajes, 1):
    # Predicción Random Forest
    pred = at.predecir_fila(row)
    
    # Filtrado vecinos: primero rango estándar
    mask = (
        (at.df['trip_distance'].between(row['trip_distance']-1.0, row['trip_distance']+1.0)) &
        (at.df['pickup_hour'].between(row['pickup_hour']-2, row['pickup_hour']+2))
    )
    neigh = at.df[mask]
    
    # Si no hay vecinos, usar rango expandido
    if len(neigh) == 0:
        mask2 = (
            (at.df['trip_distance'].between(row['trip_distance']-2.0, row['trip_distance']+2.0)) &
            (at.df['pickup_hour'].between(row['pickup_hour']-4, row['pickup_hour']+4))
        )
        neigh = at.df[mask2]
    
    if len(neigh) > 0:
        mediana_vecinos = neigh['duration_minutes'].median()
        cant_vecinos = len(neigh)
    else:
        mediana_vecinos = "No hay vecinos"
        cant_vecinos = 0
    
    resultados.append({
        'Viaje': i,
        'Predicción RF (min)': round(pred,2),
        'Vecinos encontrados': cant_vecinos,
        'Mediana vecinos (min)': round(mediana_vecinos,2) if isinstance(mediana_vecinos,float) else mediana_vecinos
    })

# Mostrar tabla final
df_resultados = pd.DataFrame(resultados)
df_resultados
```
